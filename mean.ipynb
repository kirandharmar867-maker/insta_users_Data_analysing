```python
with open("finaldata.txt", encoding="utf-8") as f:
    data = f.read()
changs=data.split("\n\n")
changs=[c for c in changs if len(c)>3]
```

```python
list_changs=[]
def parse_changs(chang):
    chang=chang.strip()
    lines=chang.split("\n")
    user_name=lines[0]
    posts=int(lines[1].split(" post")[0].replace(",", ""))
    totle_followers=float(lines[2].split(" followers")[0].replace(",", "").replace("K","").replace("M", ""))
    if ("K" in lines[2]):
        totle_followers = int(totle_followers * 1000)
    elif "M" in lines[2]:
        totle_followers = int(totle_followers * 1000000)
    else:
        totle_followers=int(totle_followers)
    totle_following=int(lines[3].split(" following")[0].replace(",", "").replace("K","").replace("M", ""))
    if ("K" in lines[3]):
        totle_following = int(totle_following * 1000)
    elif "M" in lines[3]:
        totle_following = int(totle_following * 1000000)
    else:
        totle_following=int(totle_following)
    nick_name=lines[4]
    if len(lines[5:])>3:
        type_of_page=lines[5]
        bio="\n".join(lines[6:])
    else:
        type_of_page="empty"
        bio="empty"
    return {"user_name":user_name, "posts":posts, "totle_followers":totle_followers, "totle_following":totle_following,
           "nick_name":nick_name, "type_of_page":type_of_page, "bio":bio}
    
```

```python
for chang in changs:
    parsed_changs=parse_changs(chang)
    list_changs.append(parsed_changs)
list_changs
```

```python
import json
s=json.dumps(list_changs, indent=4)
with open("final_output.json", "w")as f:
    f.write(s)
```
